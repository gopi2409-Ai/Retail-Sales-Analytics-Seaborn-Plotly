```python

import os
import pandas as pd

# Create the data folder
os.makedirs("data", exist_ok=True)

# Create the retail sales dataset
sales_data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue": [50000, 65000, 55000, 80000, 90000, 75000],
    "Expenses": [30000, 35000, 32000, 45000, 50000, 42000],
    "Orders": [120, 150, 135, 190, 210, 180]
}

df = pd.DataFrame(sales_data)

# Save the CSV file
df.to_csv("data/retail_sales.csv", index=False)

print("CSV file created successfully!")
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

# 1. Load the CSV file
df = pd.read_csv("data/retail_sales.csv")

# 2. Display the dataset
print("Dataset Preview:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

# 3. Calculate profit
df["Profit"] = df["Revenue"] - df["Expenses"]

# 4. Business summary
print("\n--- BUSINESS SUMMARY ---")
print("Total Revenue:", df["Revenue"].sum())
print("Total Expenses:", df["Expenses"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Orders:", df["Orders"].sum())
print("Average Monthly Revenue:", df["Revenue"].mean())

# 5. Revenue distribution chart
sns.histplot(data=df, x="Revenue", bins=5, kde=True)
plt.title("Monthly Revenue Distribution")
plt.xlabel("Revenue")
plt.tight_layout()
plt.show()

# 6. Revenue vs Expenses chart
sns.scatterplot(
    data=df,
    x="Expenses",
    y="Revenue",
    hue="Month",
    s=120
)
plt.title("Revenue vs Expenses")
plt.tight_layout()
plt.show()

# 7. Interactive revenue chart
fig1 = px.line(
    df,
    x="Month",
    y="Revenue",
    markers=True,
    title="Monthly Revenue Trend",
    hover_data=["Expenses", "Profit", "Orders"]
)
fig1.show()

# 8. Interactive profit chart
fig2 = px.bar(
    df,
    x="Month",
    y="Profit",
    color="Profit",
    title="Monthly Profit",
    text="Profit"
)
fig2.show()
```
